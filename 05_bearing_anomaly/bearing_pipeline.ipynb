```python
# 진동·운영 데이터 불러온 뒤
Fs = 25600            # 예시: 25.6 kHz (장비 사양 확인!)
vib_df1["t"] = np.arange(len(vib_df1)) / Fs
op_df1 ["t"] = np.arange(len(op_df1 )) / Fs   # 운영 쪽도 길이가 같다면 동일하게

# time을 인덱스로
vib_df1.set_index("t", inplace=True)
op_df1 .set_index("t", inplace=True)
```

```python
merged = vib_df1.join(op_df1, how="inner", rsuffix="_op")
# ‘inner’ = 두 DF 모두 존재하는 시점만 남김
```

```python
WIN_SEC   = 1.0        # 1초 길이로 자름 (필요하면 0.2 s, 5 s 등 실험)
STEP_SEC  = 1.0        # 겹침 없이 이동; 50% overlap이면 0.5 s

win       = int(Fs * WIN_SEC)
step      = int(Fs * STEP_SEC)

segments  = range(0, len(merged) - win + 1, step)
```

```python
def time_domain_features(arr):
    """1 채널 배열 → dict(time‑domain FIs)"""
    rms   = np.sqrt(np.mean(arr**2))
    pp    = np.ptp(arr)
    kurt  = stats.kurtosis(arr, fisher=False)
    skew  = stats.skew(arr)
    cf    = np.max(np.abs(arr)) / rms      # Crest Factor
    sh    = rms / np.mean(np.abs(arr))     # Shape Factor
    return {
        "rms": rms, "peak2peak": pp, "kurt": kurt,
        "skew": skew, "crest": cf, "shape": sh
    }

def freq_band_energy(arr, Fs, band):
    """대역별 에너지(FFT)"""
    f, Pxx = signal.welch(arr, fs=Fs, nperseg=2048)
    mask   = (f >= band[0]) & (f <= band[1])
    return np.trapz(Pxx[mask], f[mask])

def extract_fis(seg_df, Fs):
    """분할된 한 윈도우(DF) → 피처 Series"""
    feats = {}
    # ① 진동 채널별 TD features
    vib_cols = [c for c in seg_df.columns if c.startswith("Ch")]   # 예시
    for ch in vib_cols:
        td = time_domain_features(seg_df[ch].values)
        feats.update({f"{ch}_{k}": v for k, v in td.items()})
    # ② 진동 채널 FFT 대역 에너지(예: 0.5~2 kHz)
    for ch in vib_cols:
        feats[f"{ch}_E0_0.5k"] = freq_band_energy(seg_df[ch].values, Fs, (0,   500))
        feats[f"{ch}_E0.5_2k"] = freq_band_energy(seg_df[ch].values, Fs, (500, 2000))
    # ③ 운영 파라미터(평균값·최댓값 등)
    op_cols = [c for c in seg_df.columns if c.endswith("_op")]
    for oc in op_cols:
        feats[f"{oc}_mean"] = seg_df[oc].mean()
        feats[f"{oc}_max"]  = seg_df[oc].max()
    # ④ 원본 파일명(source_file) 다중 값일 경우 가장 frequent 값
    if "source_file" in seg_df:
        feats["source_file"] = seg_df["source_file"].mode().iloc[0]
    # ⑤ 윈도우 끝 시각(time index 마지막)
    feats["t_end"] = seg_df.index[-1]
    return pd.Series(feats)
```

```python
feature_rows = []
for start in segments:
    seg_df = merged.iloc[start : start + win]
    fis    = extract_fis(seg_df, Fs)
    feature_rows.append(fis)

fi_df = pd.DataFrame(feature_rows)
fi_df.reset_index(drop=True, inplace=True)
print("윈도우 수:", fi_df.shape[0], "피처 개수:", fi_df.shape[1])
```

```python
# ---------------------------------------------------------------
# 0. 환경 설정
# ---------------------------------------------------------------
import os, glob, numpy as np, pandas as pd, matplotlib.pyplot as plt
import scipy.signal as sg
from nptdms import TdmsFile
import torch, torch.nn as nn, torch.utils.data as td

ROOT_DIR   = r'/Users/qwergs/Desktop/기계데이터분석(베어링)'
ASSUMED_FS = 25600        # Hz
WIN_SEC    = 1.0          # NSP 윈도우 길이
TEMP_LIM   = 200          # °C
TORQUE_LIM = -17          # Nm

# ---------------------------------------------------------------
# A. 실제 TDMS 이름 ↔ 스크립트 내부 이름 매핑
# ---------------------------------------------------------------
NAME_MAP = {
    # 그룹
    'Vibration': 'vib',
    'Oper':      'op',
    # 채널
    'CH01':        'CH1',
    'CH02':        'CH2',
    'CH03':        'CH3',
    'CH04':        'CH4',
    'Torque':      'Torque',
    'Torque[Nm]':  'Torque',
    'Temp_Front':  'TC_Front',
    'TC SP Front[°C]': 'TC_Front',
    'Temp_Rear':   'TC_Rear',
    'TC SP Rear[°C]':  'TC_Rear',
}

def get_ch(tdms, group_key, ch_key, default=None):
    """매핑 정보를 사용해 채널 데이터를 안전하게 가져온다."""
    for g in tdms.groups():
        g_alias = NAME_MAP.get(g.name, g.name)
        if g_alias != group_key:
            continue
        for ch in g.channels():
            ch_alias = NAME_MAP.get(ch.name, ch.name)
            if ch_alias == ch_key:
                return ch[:]
    if default is not None:
        return default
    raise KeyError(f"{group_key}/{ch_key} 채널을 찾을 수 없습니다.")

# ---------------------------------------------------------------
# 1. Spectral Kurtosis 기반 공진대역 탐색
# ---------------------------------------------------------------
def top_kurtosis_bands(sig, fs, k=3, n_fft=4096):
    f, t, Sxx = sg.spectrogram(sig, fs, nperseg=n_fft, noverlap=n_fft//2)
    sk = ((Sxx - Sxx.mean(1, keepdims=True))**4).mean(1) / (Sxx.var(1)+1e-9)**2
    idx = np.argsort(sk)[-k:]
    bands = [(f[i], f[min(i+1, len(f)-1)]) for i in sorted(idx)]
    return bands

# ---------------------------------------------------------------
# 2. NSP 이미지 생성
# ---------------------------------------------------------------
def make_nsp(ch_vert, ch_vert_rear, fs):
    bands = top_kurtosis_bands(ch_vert, fs)
    img_rgb = []

    for lo, hi in bands:
        # 1) 대역 유효성 보정
        lo = max(lo, 1.0)              # DC 제거
        hi = min(hi, fs/2 - 10)        # 나이퀴스트보다 작게
        if lo >= hi:                   # 잘못된 대역이면 건너뜀
            img_rgb.append(np.zeros((128, 128)))
            continue

        # 2) 필터 & 엔벨로프
        b, a = sg.butter(4, [lo/(fs/2), hi/(fs/2)], 'band')
        env1 = np.abs(sg.hilbert(sg.filtfilt(b, a, ch_vert)))
        env2 = np.abs(sg.hilbert(sg.filtfilt(b, a, ch_vert_rear)))

        # 3) nan·inf 처리
        env1 = np.nan_to_num(env1, nan=0.0, posinf=0.0, neginf=0.0)
        env2 = np.nan_to_num(env2, nan=0.0, posinf=0.0, neginf=0.0)

        # 4) 값이 전부 0이면 빈 행렬
        if env1.max() == 0 or env2.max() == 0:
            img_rgb.append(np.zeros((128, 128)))
            continue

        # 5) 2‑D 히스토그램 → 128×128
        H, _, _ = np.histogram2d(env1, env2, bins=128,
                                 range=[[0, env1.max()], [0, env2.max()]])
        img_rgb.append((H / H.max()).T)

    # 밴드가 3개보다 적게 유효해도 항상 R‑G‑B 3채널을 맞춘다
    while len(img_rgb) < 3:
        img_rgb.append(np.zeros((128, 128)))

    img = np.stack(img_rgb[:3], -1)        # (128,128,3)
    return (img * 255).astype(np.uint8)
# ---------------------------------------------------------------
# 3. TDMS → NSP·라벨 세트 추출
# ---------------------------------------------------------------
def extract_from_tdms(tdms_path, fs, win_sec):
    tdms = TdmsFile.read(tdms_path)

    ch1 = get_ch(tdms, 'vib', 'CH1')
    ch3 = get_ch(tdms, 'vib', 'CH3')

    zero_like = np.zeros_like(ch1)
    torque   = get_ch(tdms, 'op', 'Torque',   default=zero_like)
    t_front  = get_ch(tdms, 'op', 'TC_Front', default=zero_like)
    t_rear   = get_ch(tdms, 'op', 'TC_Rear',  default=zero_like)

    win_len = int(fs * win_sec)
    n_win   = len(ch1) // win_len
    imgs, y = [], []

    fail = ((t_front >= TEMP_LIM) | (t_rear >= TEMP_LIM) |
            (torque <= TORQUE_LIM)).astype(int)
    trigger = pd.Series(fail).rolling(3).sum() >= 3
    fpt_idx = trigger.idxmax() if trigger.any() else len(fail)

    for i in range(n_win):
        s, e = i * win_len, (i + 1) * win_len
        img = make_nsp(ch1[s:e], ch3[s:e], fs)
        ratio = s / max(1, fpt_idx)
        if ratio <= 0.05:
            label = 0
        elif ratio >= 0.95:
            label = 1
        else:
            label = -1
        imgs.append(img)
        y.append(label)
    return imgs, y

# ---------------------------------------------------------------
# 4. 전체 데이터셋 구축
# ---------------------------------------------------------------
all_imgs, all_lbls = [], []
pattern = os.path.join(ROOT_DIR, 'TrainSet', 'Train*', '*.tdms')
for tdms_path in glob.glob(pattern):
    imgs, lbls = extract_from_tdms(tdms_path, ASSUMED_FS, WIN_SEC)
    all_imgs.extend(imgs)
    all_lbls.extend(lbls)

X = np.stack(all_imgs)
y = np.array(all_lbls)

idx_labeled = y >= 0
X_labeled   = X[idx_labeled] / 255.0
y_labeled   = y[idx_labeled]

# ---------------------------------------------------------------
# 5. PyTorch Dataset/Dataloader
# ---------------------------------------------------------------
class NSPDataset(td.Dataset):
    def __init__(self, X, y):
        self.X = torch.from_numpy(X).permute(0, 3, 1, 2).float()
        self.y = torch.from_numpy(y).long()
    def __len__(self):
        return len(self.y)
    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

loader = td.DataLoader(NSPDataset(X_labeled, y_labeled),
                       batch_size=32, shuffle=True)

# ---------------------------------------------------------------
# 6. CNN‑HS 모델
# ---------------------------------------------------------------
class CNNHS(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(3, 20, 10, stride=2), nn.ReLU(),      # 128 → 60
            nn.Conv2d(20, 40, 5),          nn.ReLU(),       # 60  → 56
            nn.Conv2d(40, 20, 3),          nn.ReLU(),       # 56  → 54
            nn.AdaptiveAvgPool2d((24, 24))                  # 54  → 24
        )
        self.fc = nn.Sequential(
            nn.Flatten(),                                   # 20 × 24 × 24 = 11520
            nn.Linear(11520, 500), nn.ReLU(),
            nn.Linear(500,  50),  nn.ReLU(),
            nn.Linear(50,     2)
        )
    def forward(self, x):
        return self.fc(self.conv(x))

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model  = CNNHS().to(device)
loss_fn = nn.CrossEntropyLoss()
opt     = torch.optim.Adam(model.parameters(), lr=1e-3)

# ---------------------------------------------------------------
# 7. 학습
# ---------------------------------------------------------------
for epoch in range(10):
    tot, n = 0.0, 0
    model.train()
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        opt.zero_grad()
        pred = model(xb)
        loss = loss_fn(pred, yb)
        loss.backward()
        opt.step()
        tot += loss.item() * len(yb)
        n   += len(yb)
    print(f'Epoch {epoch+1:02d}  loss {tot/n:.4f}')

# ---------------------------------------------------------------
# 8. 모델 저장
# ---------------------------------------------------------------
torch.save(model.state_dict(), 'cnn_hs_nsp.pth')
```

- 공진대역 탐색 -> NSP 이미지 생성 함수 

```python
import numpy as np, scipy.signal as sg, matplotlib.pyplot as plt
from PIL import Image
from io import BytesIO

def kurt_band(sig, fs, n_bands=3):
    """Spectral Kurtosis 기반 상위 n_bands 대역 반환"""
    f, t, Sxx = sg.spectrogram(sig, fs, nperseg=fs//2, noverlap=fs//4)
    SK = ((Sxx - Sxx.mean(axis=1,keepdims=True))**4).mean(axis=1) / (Sxx.var(axis=1)+1e-9)**2
    idx = np.argsort(SK)[-n_bands:]
    bands = [(f[i], f[i+1]) for i in idx]
    return sorted(bands)

def make_nsp(row, fs=25600):
    """CH1~4 한 윈도우(1 초 정도) → 128×128×3 NSP"""
    ch_v = row[["CH1","CH3"]].values     # x,y 좌표 (Vertical)
    ch_a = row[["CH2","CH4"]].values     # Axial → 컬러 밀도 계산용

    # (1) 대역 추출 (CH1 사용해 예시)
    bands = kurt_band(row["CH1"], fs)

    rgb = []
    for lo,hi in bands:                           # 3 대역 → RGB
        b,a  = sg.butter(4, [lo/fs*2, hi/fs*2], 'band')
        env1 = np.abs(sg.hilbert(sg.filtfilt(b,a,row["CH1"])))
        env2 = np.abs(sg.hilbert(sg.filtfilt(b,a,row["CH3"])))
        # 2d histogram → 밀도
        H,_,_ = np.histogram2d(env1, env2, bins=128, range=[[0,env1.max()],[0,env2.max()]])
        rgb.append((H/ H.max()).T)                # transpose for imshow

    img = np.stack(rgb, axis=-1)                  # (128,128,3)
    return (img*255).astype(np.uint8)
```
