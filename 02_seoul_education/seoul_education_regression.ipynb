#### 양의 상관
**< 미혼모자 공동생활가정 >**
- 이 시설이 많은 지역은 상대적으로 사회복지 시스템이 잘 작동하거나 행정/도시 기반이 잘 갖춰진 지역일 가능성
- 이런 지역일수록 교육 접근성이나 학습지원 인프라도 잘 마련되어 있어 국어 점수와 같은 학업 성취 지표가 높게 나타날 수 있음 (예: 서울시 자치구 중 종합복지계획이 잘 마련된 곳들)


**< 성폭력 피해자 보호 시설 >**
- 사회 안전망이 잘 구축된 지역에서 시설 수가 많으며, 해당 지역은 복지 예산/정책이 활발한 경우가 많음
- 간접적으로 정서적 안정, 교육 지속성 유지에 기여하는 환경이 될 가능성


**< 여성 부랑인 시설 >**
- 상관관계가 양의 방향이라는 점은 흥미로움 -> 이는 단순히 “해당 시설이 많아서 점수가 높다”기보다, 해당 시설들이 존재할 수 있는 도시적 조건 (예: 인프라, 예산, 서비스 접근성)이 국어 점수와도 관련된 것
- 복지 수요뿐 아니라 복지 공급능력이 있는 지역일 가능성

#### 음의 상관
**< 자립 지원 시설 >**
- 보호종료 아동(예: 아동복지시설 퇴소자)의 자립을 돕는 거주시설
- 취약계층의 비율이 높은 지역에 설치되며, 사회적 불안정성, 소득 수준, 가정 내 교육환경 부족과 연결
- 해당 지역의 학생들은 학습 지속성/동기/정서적 지원 부족으로 국어 성취도에 부정적 영향 가능

**< 가정폭력 피해자 지원 시설 >**
- 이 시설이 많다는 것은 해당 지역에서 가정 내 불안정성이 상대적으로 높다는 신호
- 이는 아동의 학습 환경 불안정 -> 집중력/정서 불안 -> 학업 성취도 하락이라는 연쇄로 이어질 수 있음

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

def compute_vif(df):
    vif_data = pd.DataFrame()
    vif_data["feature"] = df.columns
    vif_data["VIF"] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
    return vif_data

targets = ['영어척도점수평균점수', '수학척도점수평균점수', '국어척도점수평균점수']

X = df.drop(columns=targets)  # 종속변수 제거
X = X.dropna()  # 결측값 제거 (필수)

vif_result = compute_vif(X)
print(vif_result.sort_values(by='VIF', ascending=False))
```

```python
# 제외할 키워드를 리스트로 정의
exclude_keywords = ['인원수', '비율', '성폭력', '총전입 (명)', '총전출 (명)', '시도내이동-시군구간 전입 (명)',
                    '시도내이동-시군구간 전출 (명)', '시도내이동-시군구내 (명)', '소계', '합계', '전출']

# 변수명 중 키워드를 포함하는 열만 추출하여 리스트로 저장
exclude_cols = [col for col in df.columns if any(keyword in col for keyword in exclude_keywords)]

# 해당 열들을 DataFrame에서 제거
df_filtered = df.drop(columns=exclude_cols)

X = df_filtered.drop(columns=targets)  # 종속변수 제거
X = X.dropna()  # 결측값 제거 (필수)

vif_result = compute_vif(X)
print(vif_result.sort_values(by='VIF', ascending=False))
```

```python
# 후진 제거법

from statsmodels.api import OLS, add_constant

# 후진제거법 함수 정의
def backward_elimination(X, y, threshold_out=0.051, verbose=True):
    X = add_constant(X)
    included = list(X.columns)

    while True:
        model = OLS(y, X[included]).fit()
        pvalues = model.pvalues.drop('const', errors='ignore')
        max_pval = pvalues.max()

        if max_pval > threshold_out:
            worst_feature = pvalues.idxmax()
            included.remove(worst_feature)
            if verbose:
                print(f"제거: {worst_feature} (p = {max_pval:.4f})")
        else:
            break

    return [var for var in included if var != 'const']


# 종속변수 목록
targets = ['영어척도점수평균점수', '수학척도점수평균점수', '국어척도점수평균점수']

# 결과 저장
selected_vars_dict = {}

for target in targets:
    print(f"\n===== [{target}] 변수 선택 결과 =====")
    # y 설정
    y = df[target]

    # X와 y 결합하여 결측값 제거
    data = pd.concat([X, y], axis=1).dropna()
    X_clean = data.drop(columns=[target])
    y_clean = data[target]

    # 후진제거법 실행
    selected_vars = backward_elimination(X_clean, y_clean, verbose=True)
    selected_vars_dict[target] = selected_vars

    print("✅ 최종 선택된 변수:")
    print(selected_vars)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cross_decomposition import CCA
from scipy.stats import chi2

# X, Y가 DataFrame이 아닌 경우, 열 이름을 임의로 지정
X_columns = [f'X{i+1}' for i in range(X.shape[1])]
Y_columns = [f'Y{i+1}' for i in range(Y.shape[1])]

# ✅ 1. 표준화
X_std = StandardScaler().fit_transform(X)
Y_std = StandardScaler().fit_transform(Y)

# ✅ 2. 정준상관분석 수행
n_components = min(X.shape[1], Y.shape[1])
cca = CCA(n_components=n_components)
cca.fit(X_std, Y_std)
X_c, Y_c = cca.transform(X_std, Y_std)

# ✅ 3. Wilks’ Lambda 유의성 검정 함수 정의
def wilks_lambda_test(X_c, Y_c, p, q, n, n_components):
    wilks_lambdas = []
    chisq_vals = []
    df_vals = []
    p_vals = []

    for i in range(n_components):
        r2 = np.corrcoef(X_c[:, i], Y_c[:, i])[0, 1] ** 2
        wilks = np.prod([1 - r2])
        df = (p - i) * (q - i)
        chisq = -(n - 1 - (p + q + 1) / 2) * np.log(wilks)
        p_val = 1 - chi2.cdf(chisq, df)

        wilks_lambdas.append(wilks)
        chisq_vals.append(chisq)
        df_vals.append(df)
        p_vals.append(p_val)

    return pd.DataFrame({
        "Wilks' Lambda": wilks_lambdas,
        "Chi-Square": chisq_vals,
        "df": df_vals,
        "p-value": p_vals
    })

# ✅ 4. 유의성 검정 실행
n = X_std.shape[0]
p = X_std.shape[1]
q = Y_std.shape[1]
wilks_results = wilks_lambda_test(X_c, Y_c, p, q, n, n_components)
print("✅ Wilks’ Lambda 유의성 검정 결과:")
print(wilks_results)

# ✅ 5. 정준부하량 계산 및 출력
x_loadings = np.corrcoef(X_std.T, X_c.T)[:p, p:]
y_loadings = np.corrcoef(Y_std.T, Y_c.T)[:q, q:]

x_loadings_df = pd.DataFrame(x_loadings, index=X_columns, columns=[f'U{i+1}' for i in range(n_components)])
y_loadings_df = pd.DataFrame(y_loadings, index=Y_columns, columns=[f'V{i+1}' for i in range(n_components)])

print("\n✅ X 정준부하량:")
print(x_loadings_df)

print("\n✅ Y 정준부하량:")
print(y_loadings_df)
```
